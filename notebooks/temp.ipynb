```python
import pandas as pd
from scipy.stats import hypergeom, binom

def p_draw_at_least_x(population_size: int, population_successes: int, sample_size: int, desired_successes: int) -> float:
    return 1 - float(hypergeom.cdf(desired_successes-1, population_size, population_successes, sample_size))

def p_draw_with_x_mulligans(num_mulligans: int, p: float) -> float:
    return 1 - float(binom.cdf(0, num_mulligans, p))

def num_hands_as_column_name(i: int | str) -> str:
    if isinstance(i, str):
        return i
    return "opener" if i == 1 else f'{i-1} mulligan(s)'

def p_num_lands_in_opening_hand(num_lands_in_opener: int, min_lands: int = 30, max_lands: int = 40, num_hands: int = 4) -> pd.DataFrame:
    prob_dict = {} 
    prob_dict['num_lands'] = [x for x in range(min_lands, max_lands + 1)]
    for num_hands in range(1, num_hands + 1):
        prob_dict[num_hands] = []
        for num_lands in range(min_lands, max_lands + 1):
            p_draw_in_one_hand = p_draw_at_least_x(99, num_lands, 7, num_lands_in_opener)
            p_draw_in_x_mulls = p_draw_with_x_mulligans(num_hands, p_draw_in_one_hand)
            prob_dict.get(num_hands).append(p_draw_in_x_mulls)
    df = pd.DataFrame(prob_dict)
    df.columns = [num_hands_as_column_name(x) for x in df.columns]
    return df

p_num_lands_in_opening_hand(3).head(99).style.hide()
```
